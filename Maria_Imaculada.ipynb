```python
import numpy as np
A = np.array([[2, 1, 1],
              [3, 5, 2],
              [1, 2, 4]], dtype=float)

b = np.array([8, 18, 15], dtype=float)
solusi = np.linalg.solve(A, b)

print("Nilai x, y, dan z secara berurutan adalah:")
print(f"x = {solusi[0]:.2f}")
print(f"y = {solusi[1]:.2f}")
print(f"z = {solusi[2]:.2f}")
```
