```python

# ============================================
# PROGRAM PENYELESAIAN SPLTV
# METODE ELIMINASI GAUSS
# ============================================

# Fungsi untuk menyelesaikan sistem persamaan
def eliminasi_gauss(A, B):

    n = len(B)

    # Membuat matriks augmented
    M = [A[i][:] + [B[i]] for i in range(n)]

    # Proses eliminasi Gauss
    for i in range(n):

        # Mencari pivot terbesar
        pivot = max(range(i, n), key=lambda r: abs(M[r][i]))

        # Menukar baris jika diperlukan
        M[i], M[pivot] = M[pivot], M[i]

        # Memeriksa pivot nol
        if abs(M[i][i]) < 1e-10:
            continue

        # Membuat elemen diagonal menjadi 1
        pembagi = M[i][i]
        for j in range(i, n + 1):
            M[i][j] /= pembagi

        # Menghilangkan elemen di bawah pivot
        for k in range(i + 1, n):
            faktor = M[k][i]
            for j in range(i, n + 1):
                M[k][j] -= faktor * M[i][j]

    # Memeriksa apakah sistem memiliki solusi
    for i in range(n):
        if all(abs(M[i][j]) < 1e-10 for j in range(n)):
            if abs(M[i][n]) >= 1e-10:
                print("Sistem tidak memiliki solusi.")
                return

    # Substitusi mundur
    x = [0.0] * n

    for i in range(n - 1, -1, -1):
        if abs(M[i][i]) < 1e-10:
            print("Sistem memiliki banyak solusi.")
            return

        jumlah = sum(M[i][j] * x[j]
                     for j in range(i + 1, n))

        x[i] = M[i][n] - jumlah

    # Menampilkan hasil
    print("\nHASIL PENYELESAIAN:")
    print("Nilai x =", round(x[0], 4))
    print("Nilai y =", round(x[1], 4))
    print("Nilai z =", round(x[2], 4))


# ============================================
# INPUT DATA PERSAMAAN
# ============================================

print("PROGRAM SPLTV METODE ELIMINASI GAUSS")
print("------------------------------------")

A = []
B = []

for i in range(3):
    print("\nPersamaan ke-", i + 1)

    a = float(input("Masukkan koefisien x: "))
    b = float(input("Masukkan koefisien y: "))
    c = float(input("Masukkan koefisien z: "))
    d = float(input("Masukkan nilai konstanta: "))

    A.append([a, b, c])
    B.append(d)

# Memanggil fungsi eliminasi Gauss
eliminasi_gauss(A, B)

```
