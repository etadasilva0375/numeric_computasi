```python

# ==================================================
# PROGRAM PENYELESAIAN SPLTV
# MENGGUNAKAN METODE ELIMINASI GAUSS
# ==================================================

# Fungsi untuk menyelesaikan sistem persamaan linear
# A adalah matriks koefisien, sedangkan B adalah konstanta.
def eliminasi_gauss(A, B):

    # Menghitung jumlah persamaan atau variabel.
    n = len(B)

    # Membuat matriks augmented dengan menggabungkan
    # matriks koefisien A dan konstanta B.
    M = [A[i][:] + [B[i]] for i in range(n)]

    # Menentukan batas toleransi untuk bilangan nol.
    toleransi = 1e-10

    # ==============================================
    # PROSES ELIMINASI GAUSS
    # ==============================================

    # Melakukan eliminasi pada setiap kolom matriks.
    for i in range(n):

        # Mencari baris dengan nilai pivot terbesar
        # agar perhitungan lebih stabil.
        pivot = max(range(i, n),
                    key=lambda r: abs(M[r][i]))

        # Menukar baris jika posisi pivot berbeda.
        M[i], M[pivot] = M[pivot], M[i]

        # Memeriksa apakah nilai pivot mendekati nol.
        if abs(M[i][i]) < toleransi:
            continue

        # Membagi baris pivot agar elemen diagonal
        # utama bernilai satu.
        pembagi = M[i][i]

        for j in range(i, n + 1):
            M[i][j] = M[i][j] / pembagi

        # Menghilangkan elemen di bawah pivot
        # menggunakan operasi baris elementer.
        for k in range(i + 1, n):

            # Menentukan faktor pengali eliminasi.
            faktor = M[k][i]

            # Mengurangi elemen baris dengan baris pivot.
            for j in range(i, n + 1):
                M[k][j] = M[k][j] - faktor * M[i][j]

    # ==============================================
    # MEMERIKSA KEBERADAAN SOLUSI
    # ==============================================

    # Memeriksa apakah terdapat persamaan yang
    # tidak memiliki solusi.
    for i in range(n):

        # Jika semua koefisien nol tetapi konstanta
        # bukan nol, sistem tidak memiliki solusi.
        if all(abs(M[i][j]) < toleransi
               for j in range(n)):

            if abs(M[i][n]) >= toleransi:
                print("Sistem tidak memiliki solusi.")
                return

    # Memeriksa apakah sistem memiliki banyak solusi.
    for i in range(n):
        if abs(M[i][i]) < toleransi:
            print("Sistem memiliki banyak solusi.")
            return

    # ==============================================
    # PROSES SUBSTITUSI MUNDUR
    # ==============================================

    # Menyiapkan tempat untuk menyimpan nilai variabel.
    x = [0.0] * n

    # Menghitung nilai variabel mulai dari baris terakhir.
    for i in range(n - 1, -1, -1):

        # Menghitung jumlah hasil perkalian
        # koefisien dengan variabel yang sudah diketahui.
        jumlah = sum(M[i][j] * x[j]
                     for j in range(i + 1, n))

        # Menghitung nilai variabel saat ini.
        x[i] = M[i][n] - jumlah

    # ==============================================
    # MENAMPILKAN HASIL PERHITUNGAN
    # ==============================================

    # Menampilkan judul hasil penyelesaian.
    print("\nHASIL PENYELESAIAN SPLTV")

    # Menampilkan nilai masing-masing variabel.
    for i in range(n):
        print(f"Nilai variabel ke-{i+1} = {x[i]:.4f}")


# ==================================================
# BAGIAN INPUT DATA
# ==================================================

# Menampilkan judul program.
print("PROGRAM SPLTV METODE ELIMINASI GAUSS")
print("------------------------------------")

# Menyiapkan daftar untuk menyimpan koefisien.
A = []

# Menyiapkan daftar untuk menyimpan konstanta.
B = []

# Melakukan input sebanyak tiga persamaan.
for i in range(3):

    # Menampilkan nomor persamaan.
    print(f"\nPersamaan ke-{i+1}")

    # Memasukkan koefisien variabel x.
    a = float(input("Masukkan koefisien x: "))

    # Memasukkan koefisien variabel y.
    b = float(input("Masukkan koefisien y: "))

    # Memasukkan koefisien variabel z.
    c = float(input("Masukkan koefisien z: "))

    # Memasukkan nilai konstanta persamaan.
    d = float(input("Masukkan nilai konstanta: "))

    # Menyimpan koefisien ke dalam matriks A.
    A.append([a, b, c])

    # Menyimpan konstanta ke dalam matriks B.
    B.append(d)

# Memanggil fungsi untuk menyelesaikan SPLTV.
eliminasi_gauss(A, B)
```

PROGRAM PENYELESAIAN SISTEM PERSAMAAN
LINEAR TIGA VARIABEL (SPLTV)

Metode: Eliminasi Gauss

Tujuan:
Program ini dibuat untuk menyelesaikan sistem
persamaan linear tiga variabel, yaitu x, y, dan z,
menggunakan metode Eliminasi Gauss.

Metode ini bekerja dengan mengubah persamaan
menjadi matriks augmented, melakukan operasi
baris elementer, kemudian mencari nilai variabel
menggunakan substitusi mundur.

PENJELASAN PROGRAM

1. Input Data
   Pengguna memasukkan koefisien x, y, z dan
   konstanta dari tiga persamaan.

2. Matriks Augmented
   Koefisien dan konstanta digabungkan menjadi
   satu matriks untuk mempermudah perhitungan.

3. Eliminasi Gauss
   Program mengubah elemen di bawah diagonal
   utama menjadi nol menggunakan operasi
   baris elementer.

4. Substitusi Mundur
   Program menghitung nilai variabel mulai dari
   variabel terakhir sampai variabel pertama.

5. Output
   Program menampilkan hasil nilai x, y, dan z.

KESIMPULAN

Program Python ini dapat digunakan untuk
menyelesaikan Sistem Persamaan Linear Tiga
Variabel menggunakan metode Eliminasi Gauss.
Setiap tahapan dilakukan secara sistematis
mulai dari input data hingga hasil akhir.
