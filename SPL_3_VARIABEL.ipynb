```python
import numpy as np

A = np.array([
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
], dtype=float)

b = np.array([8, -11, -3], dtype=float)

print("Matriks A:")
print(A)

print("Konstanta b:")
print(b)

aug = np.column_stack((A, b))

print("Matriks Augmented:")
print(aug)

n = len(b)

for kolom in range(n):

    baris_pivot = kolom + np.argmax(
        np.abs(aug[kolom:, kolom])
    )

    if baris_pivot != kolom:
        aug[[kolom, baris_pivot]] = aug[[baris_pivot, kolom]]

    if np.isclose(aug[kolom, kolom], 0.0):
        raise ValueError("Sistem tidak memiliki solusi tunggal.")

    for baris in range(kolom + 1, n):
        faktor = aug[baris, kolom] / aug[kolom, kolom]

        aug[baris, kolom:] = (
            aug[baris, kolom:] - faktor * aug[kolom, kolom:]
        )

print("HASIL PENYELESAIAN SPL 3 VARIABEL")

print(f"Nilai x = {x_hasil[0]:.2f}")
print(f"Nilai y = {x_hasil[1]:.2f}")
print(f"Nilai z = {x_hasil[2]:.2f}")

print("\nPEMERIKSAAN HASIL:")
print(A @ x_hasil)

print("Apakah hasil sesuai?",
      np.allclose(A @ x_hasil, b))
```
